3_load_csv

```python

import pandas as pd
DATA_PATH = "/content/data/SuperMarket Analysis.csv"
df = pd.read_csv(DATA_PATH)
df.head()
```

**4 preprocess fixed**

Removes extra spaces from column names.

Converts all column names to lowercase.

```python

# clean column names
df.columns = [c.strip().lower() for c in df.columns]

# rename relevant columns
rename_map = {
    'cogs': 'expenses',
    'gross income': 'profit',
    'product line': 'product_line',
    'gross margin percentage': 'gross_margin_percent',
    'invoice id': 'invoice_id'
}
df.rename(columns=rename_map, inplace=True)

# add or derive required columns if missing
if 'expenses' not in df.columns and 'cogs' in df.columns:
    df['expenses'] = df['cogs']

if 'sales' not in df.columns and 'total' in df.columns:
    df['sales'] = df['total']

if 'profit' not in df.columns:
    df['profit'] = df['sales'] - df['expenses']

# convert 'date' to datetime and extract month
df['date'] = pd.to_datetime(df['date'])
df['month'] = df['date'].dt.to_period('M').astype(str)

# approximate customers per month
monthly_customers = df.groupby('month')['invoice_id'].nunique()

# aggregate numeric data by month
monthly_df = df.groupby('month').agg({
    'sales': 'sum',
    'expenses': 'sum',
    'profit': 'sum'
}).reset_index()

monthly_df['customers'] = monthly_df['month'].map(monthly_customers)

# fill dummy values for columns not in dataset
monthly_df['inventory_cost'] = 0
monthly_df['marketing_spend'] = 0

# ensure numeric
for col in ["sales", "expenses", "customers", "inventory_cost", "marketing_spend"]:
    monthly_df[col] = pd.to_numeric(monthly_df[col], errors="coerce").fillna(0)

# derived field
monthly_df['profit'] = monthly_df['sales'] - monthly_df['expenses']
monthly_df['month_str'] = monthly_df['month'].astype(str)

# build text records
records = []
for _, row in monthly_df.iterrows():
    text = (f"Month: {row['month_str']}. "
            f"Sales: ₹{int(row['sales']):,}. "
            f"Expenses: ₹{int(row['expenses']):,}. "
            f"Profit: ₹{int(row['profit']):,}. "
            f"Customers: {int(row['customers'])}. "
            f"Inventory Cost: ₹{int(row['inventory_cost']):,}. "
            f"Marketing Spend: ₹{int(row['marketing_spend']):,}.")
    records.append({"id": row['month_str'], "text": text, "meta": row.to_dict()})

print(f"✅ Created {len(records)} records")
print(records[:2])
```

 5 embeddings chroma

 SentenceTransformer: creates dense vector embeddings for text.
 chromadb: local vector database for storing/querying those embeddings.

```python

from sentence_transformers import SentenceTransformer
import chromadb

# Load embedding model
embed_model = SentenceTransformer('all-MiniLM-L6-v2')  # small & fast

# Initialize Chroma client (new API)
# You can store in memory or persist to disk
client = chromadb.PersistentClient(path="/content/chroma_db")

collection_name = "sme_records"

# Delete collection if exists (safe reset)
try:
    client.delete_collection(collection_name)
except Exception:
    pass

# Create new collection
collection = client.create_collection(name=collection_name)

# Prepare data
texts = [r["text"] for r in records]
metas = [r["meta"] for r in records]
ids = [r["id"] for r in records]

# Create embeddings
embeddings = embed_model.encode(texts, show_progress_bar=True, convert_to_numpy=True)

# Add to Chroma collection
collection.add(
    ids=ids,
    documents=texts,
    metadatas=metas,
    embeddings=embeddings.tolist()
)

print(f"✅ Indexed {len(ids)} records successfully in Chroma!")
```

6 retriever

k: number of most relevant results you want to retrieve (default = 4).

.tolist() converts the NumPy array into a plain Python list — Chroma expects standard lists, not NumPy arrays.

```python

def retrieve_relevant(query, k=4):
    # embed query
    q_emb = embed_model.encode([query], convert_to_numpy=True)[0].tolist()
    results = collection.query(query_embeddings=[q_emb], n_results=k, include=['documents','metadatas','distances'])
    # structure
    res = []
    for doc,meta,dist in zip(results['documents'][0], results['metadatas'][0], results['distances'][0]):
        res.append({"text": doc, "meta": meta, "distance": dist})
    return res

# quick test:
print(retrieve_relevant("Profit in May 2023", k=3))
```

```python
# 7_generator_setup
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, pipeline
GEN_MODEL = "google/flan-t5-small"  # small and works offline
tokenizer = AutoTokenizer.from_pretrained(GEN_MODEL)
model = AutoModelForSeq2SeqLM.from_pretrained(GEN_MODEL)
generator = pipeline("text2text-generation", model=model, tokenizer=tokenizer, max_length=256)
```

# 8_prompting

```python
# 8_prompting
from textwrap import dedent

def build_rag_prompt(query, retrieved, instructions="Answer briefly and clearly. Use ₹ for rupees."):
    # Build context: include top retrieved docs (text)
    context = "\n\n".join([f"- {r['text']}" for r in retrieved])
    prompt = dedent(f"""
    You are a helpful SME business analyst assistant.
    {instructions}

    Context (from the business data):
    {context}

    Question: {query}

    Provide:
    1) A short answer (one line).
    2) A brief explanation (2-3 lines) referencing the context when necessary.
    """)
    return prompt

# Quick test
q = "What was the profit in May 2023?"
ret = retrieve_relevant(q, k=4)
prompt = build_rag_prompt(q, ret)
print(prompt)
```

```python
# 9_rag_query
import re

def get_numeric_from_retrieved(retrieved, field_name):
    """
    Attempt to find numeric value for a field_name (sales, expenses, profit) from retrieved metadata.
    Returns first match.
    """
    for r in retrieved:
        meta = r.get("meta", {})
        # keys in meta might be different; try to match
        for k,v in meta.items():
            if field_name in str(k).lower() and pd.notna(v):
                return v
    return None

def rag_query(query, k=4):
    # 1. retrieve
    retrieved = retrieve_relevant(query, k=k)
    # 2. rule-based numeric extraction for simple queries (profit/sales/expenses)
    qlow = query.lower()
    # check for month mentions (like "May 2023" -> "May-23")
    # naive extraction:
    months = [r["meta"].get("month") for r in retrieved if r["meta"].get("month")]
    # attempt to extract month token from query
    month_match = re.search(r'(jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)[\w\-\s]*\d{2,4}', qlow, re.I)
    explicit_month = None
    if month_match:
        explicit_month = month_match.group(0)
        # unify: transform "may 2023" -> "May-23" maybe, but we compare loosely
    # Simple numeric answers:
    if "profit" in qlow:
        # try to find exact profit by inspecting retrieved metas or text
        val = get_numeric_from_retrieved(retrieved, "profit")
        if val is not None:
            ans_line = f"Profit = ₹{int(val):,}."
            explanation = "I found the profit directly from the data record."
            return {"answer": ans_line, "explanation": explanation, "sources": retrieved}
    # If not handled numerically, fallback to generator
    prompt = build_rag_prompt(query, retrieved)
    gen = generator(prompt, max_length=200, do_sample=False)[0]["generated_text"]
    return {"answer": gen.strip(), "explanation": None, "sources": retrieved}

# Quick test
print(rag_query("What was the profit in May 2023?"))
print(rag_query("Summarize business performance in Q1 2023."))
print(rag_query("Suggest 2 strategies to improve profits in June 2023."))
```

```python
# 10_recommendations_tool
def simple_recommendations_for_month(month_str, top_n=2):
    # find the row
    row = df[df["month"].astype(str).str.contains(month_str, case=False)]
    if row.empty:
        # fall back to rag
        resp = rag_query(f"Suggest {top_n} strategies to improve profits in {month_str}.")
        return resp["answer"]
    r = row.iloc[0]
    recs = []
    # heuristics: inventory_cost high relative to 6-month median?
    if r["inventory_cost"] >= df["inventory_cost"].rolling(window=min(6,len(df))).median().iloc[-1]:
        recs.append("Reduce inventory holding costs: consider JIT ordering or negotiate better supplier terms.")
    if r["marketing_spend"] < df["marketing_spend"].median():
        recs.append("Increase targeted marketing spend to regain lost customers (digital ads or promotions).")
    if r["expenses"] > df["expenses"].mean():
        recs.append("Review operational expenses for cost-saving: staff schedules, utility usage, renegotiate services.")
    if not recs:
        # fallback to model
        resp = rag_query(f"Suggest {top_n} strategies to improve profits in {month_str}.")
        return resp["answer"]
    return "1. " + "\n2. ".join(recs[:top_n])

# quick test
print(simple_recommendations_for_month("Jun-23"))
```

```python
# 11_chatbot_wrapper
def chatbot_response(user_query):
    res = rag_query(user_query, k=4)
    # prepare a user-friendly output
    answer = res["answer"]
    # include top sources (month ids)
    sources = [s["meta"].get("month") for s in res["sources"]]
    sources_text = " | ".join([str(s) for s in sources if s])
    return f"{answer}\n\nSources: {sources_text}"

# quick manual test
print(chatbot_response("What was profit in May-23?"))
print(chatbot_response("How were sales in Q1 2023?"))
```
